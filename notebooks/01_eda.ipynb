```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
file_path = "../data/raw/yellow_tripdata_2025-01.parquet"

df = pd.read_parquet(file_path)

print("Shape:", df.shape)

df.head()
```

```python
df.columns.tolist()
```

```python
df.info()
```

```python
df.describe(include="all").T
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nDuplicates:")
print(df.duplicated().sum())

print("\nMissing values:")
print(
    df.isnull()
      .sum()
      .sort_values(ascending=False)
)
```

```python
missing = (
    df.isnull()
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

missing
```

```python
numeric_cols = [
    "passenger_count",
    "trip_distance",
    "fare_amount",
    "tip_amount",
    "tolls_amount",
    "total_amount"
]

df[numeric_cols].describe(
    percentiles=[0.01, 0.05, 0.5, 0.95, 0.99]
).T
```

```python
df["trip_duration_minutes"] = (
    df["tpep_dropoff_datetime"]
    - df["tpep_pickup_datetime"]
).dt.total_seconds() / 60
```

```python
df["trip_duration_minutes"].describe(
    percentiles=[0.01, 0.05, 0.5, 0.95, 0.99]
)
```

```python
print(
    "Zero/negative duration:",
    (df["trip_duration_minutes"] <= 0).sum()
)

print(
    "Trips > 3 hours:",
    (df["trip_duration_minutes"] > 180).sum()
)
```

```python
df["avg_speed_mph"] = (
    df["trip_distance"] /
    (df["trip_duration_minutes"] / 60)
)
```

```python
df["avg_speed_mph"].describe(
    percentiles=[0.50, 0.90, 0.95, 0.99, 0.999]
)
```

```python
print(
    "Speed > 80 mph:",
    (df["avg_speed_mph"] > 80).sum()
)
```

```python
zones = pd.read_csv(
    "../data/raw/taxi_zone_lookup.csv"
)

zones.head()
```

```python
zones.shape
```

```python
zones["Borough"].value_counts()
```

```python
test = df[
    [
        "PULocationID",
        "DOLocationID",
        "tpep_pickup_datetime",
        "trip_distance",
        "total_amount"
    ]
].head(1000)
```

```python
test = test.merge(
    zones[["LocationID", "Borough", "Zone"]],
    left_on="PULocationID",
    right_on="LocationID",
    how="left"
)
```

```python
test = test.rename(
    columns={
        "Borough": "pickup_borough",
        "Zone": "pickup_zone"
    }
)

test[
    [
        "PULocationID",
        "pickup_borough",
        "pickup_zone"
    ]
].head(20)
```

```python
quick_demand = (
    test.groupby("pickup_zone")
        .size()
        .sort_values(ascending=False)
        .head(10)
)

quick_demand
```
